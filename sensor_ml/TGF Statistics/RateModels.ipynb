```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
"""
Demonstrate Poisson, Exponential Distributions applicability. Instantaneous linearization of the lognormal distribution
Poisson: Density of events given countrate
Exponential: Probability of Peak pileup, tail pileup
Lognormal: Model of

"""
```

```python
from numpy.random import Generator, PCG64
rng = Generator(PCG64())

plt.figure(figsize=(5,5), dpi=200)

for lam in [1,2,3,4,5]:
    edges = np.arange(0, 20)
    hist, edges = np.histogram(rng.poisson(lam=lam, size=100000), bins=edges, density=True)
    bins = edges[:-1] + np.diff(edges)
    
    plt.plot(bins, hist, marker='.', label='Lambda: {}'.format(lam))

plt.legend()
plt.title("Poisson")
plt.xlabel('k: Expected Number of Occurences on Interval')
plt.ylabel('P(X=k)')
```
